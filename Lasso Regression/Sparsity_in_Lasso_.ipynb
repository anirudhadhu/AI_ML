Why Lasso regression creates sparsity ? It means some values becomes 0 why ?

- Lasso regression uses an L1 penalty on coefficients, and the geometry of that penalty makes it cheaper (in terms of loss) to set some coefficients exactly to zero rather than just making them small. That’s why Lasso tends to produce sparse models.

For instance,

You have:

many toys (these are the features)

but the backpack is small (this is the model constraint)

Your parent says:

“You can only take toys whose total weight is small.”

Now — here’s the difference:

🟦 Ridge (no sparsity): “Make everything lighter”

Parent says:

“Every toy must lose a little bit of weight.”

So every toy becomes smaller — but you still carry all of them.

That’s Ridge.

🟨 Lasso (sparsity): “Leave some toys at home”

Parent says:

“Instead of shrinking every toy, it’s better to just not take some.”

So you look at your toys and think:

Favorite toy? Keep it!

Kind of useless toy? Leave it at home.

In the end:

✔ You carry fewer toys
✔ Some toys are gone completely (weight becomes zero)

That’s Lasso.

⭐ The key idea

Lasso prefers to say:

“This toy is not important — let’s take zero of it.”

instead of:

“Keep a tiny bit of every toy.”

So Lasso creates sparsity because it chooses a few important things and ignores the rest.
