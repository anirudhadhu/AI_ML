What is ElasticNet?

ElasticNet is a linear regression model that combines:

L1 penalty (Lasso) → encourages sparsity (feature selection)

L2 penalty (Ridge) → shrinks coefficients to reduce variance

It’s useful when:

You have many correlated features

Pure Lasso is unstable (randomly picks one among correlated features)

You want both regularization + feature selection

When to use ElasticNet

Use it when:

✔ features are highly correlated
✔ you want some coefficients exactly zero
✔ you want stable predictions
✔ you want better generalization than plain linear regression

Avoid if:

✘ data is very small and simple
✘ interpretability must be perfectly clear (regularization biases coefficients)

```python

from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python

# Ridge
reg = Ridge(alpha=0.1)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python

# Lasso
reg = Lasso(alpha=0.01)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python

# ElasticNet
reg = ElasticNet(alpha=0.005,l1_ratio=0.9)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```
