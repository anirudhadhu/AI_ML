# MICE Stands for Multivariate Imputation by Chained Equations



Three categories of missing data

1. MCAR (Missing completely at random)
2. MAR (Missing t Random)
3. MNAR (Missing not at Random)


# Adv and Disadv

Advantage

1. Accurate

Disadvanntage

1. Slow
2. Memory


How does MICE Work ?

**Stage 0**

1. Fill all the NaN values with mean of the respective columns.

**Stage 1; Shifting process **

2. Remove all col1 missing values
3. Predict the missing values of col1 using other cols

 Again

4. Remove col2 missing values
5. Predict the missing values of col2 using other cols

**Stage 2**

1. We subract the value from itration 1 (stage 1) from 0 (stage 0) we do this until and unless the value is 0 or near 0


```python

import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
```

```python

df = np.round(pd.read_csv('50_Startups.csv')[['R&D Spend','Administration','Marketing Spend','Profit']]/10000)
np.random.seed(9)
df = df.sample(5)
df
```

```python
df = df.iloc[:,0:-1]
df
```

```python

df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan
```

```python

df.head()
```

```python
# Step 1 - Impute all missing values with mean of respective col

df0 = pd.DataFrame()

df0['R&D Spend'] = df['R&D Spend'].fillna(df['R&D Spend'].mean())
df0['Administration'] = df['Administration'].fillna(df['Administration'].mean())
df0['Marketing Spend'] = df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
```

```python

# 0th Iteration
df0
```

```python

# Remove the col1 imputed value
df1 = df0.copy()

df1.iloc[1,0] = np.nan

df1
```

```python

# Use first 3 rows to build a model and use the last for prediction

X = df1.iloc[[0,2,3,4],1:3]
X
```

```python

y = df1.iloc[[0,2,3,4],0]
y
```

```python

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[1,1:].values.reshape(1,2))
```

```python

df1.iloc[1,0] = 23.14
```

```python

df1
```

```python

# Remove the col2 imputed value

df1.iloc[3,1] = np.nan

df1
```

```python

# Use last 3 rows to build a model and use the first for prediction
X = df1.iloc[[0,1,2,4],[0,2]]
X
```

```python

y = df1.iloc[[0,1,2,4],1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[3,[0,2]].values.reshape(1,2))
```

```python

df1.iloc[3,1] = 11.06
```

```python

df1
```

```python
# Remove the col3 imputed value
df1.iloc[4,-1] = np.nan

df1
```

```python

# Use last 3 rows to build a model and use the first for prediction
X = df1.iloc[0:4,0:2]
X
```

```python
y = df1.iloc[0:4,-1]
y
```

```python

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[4,0:2].values.reshape(1,2))
```

```python

df1.iloc[4,-1] = 31.56
```

```python
# After 1st Iteration
df1
```

```python

# Subtract 0th iteration from 1st iteration

df1 - df0
```

```python

df2 = df1.copy()

df2.iloc[1,0] = np.nan

df2
```

```python
X = df2.iloc[[0,2,3,4],1:3]
y = df2.iloc[[0,2,3,4],0]

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df2.iloc[1,1:].values.reshape(1,2))
```

```python

df2.iloc[1,0] = 23.78
```

```python

df2.iloc[3,1] = np.nan
X = df2.iloc[[0,1,2,4],[0,2]]
y = df2.iloc[[0,1,2,4],1]

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df2.iloc[3,[0,2]].values.reshape(1,2))
```

```python

df2.iloc[3,1] = 11.22
```

```python

df2.iloc[4,-1] = np.nan

X = df2.iloc[0:4,0:2]
y = df2.iloc[0:4,-1]

lr = LinearRegression()
lr.fit(X,y)
lr.predict(df2.iloc[4,0:2].values.reshape(1,2))
```

```python

df2.iloc[4,-1] = 31.56
```
